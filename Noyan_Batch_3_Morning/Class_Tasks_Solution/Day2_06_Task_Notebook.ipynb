## Task 1 — Three Customers

Create variables for **three fictional customers**, each with:
- `tenure_months` (an `int`)
- `monthly_charges` (a `float`)

Name them clearly, e.g. `customer1_tenure`, `customer1_charges`, etc. (or use three small dictionaries if you completed that pattern in class).


````python
# Customer 1
customer1_tenure = 24          # int: months with the company
customer1_charges = 85.50      # float: monthly charges in dollars

# Customer 2
customer2_tenure = 6           # int
customer2_charges = 45.75      # float

# Customer 3
customer3_tenure = 18          # int
customer3_charges = 65.00      # float

print(customer1_tenure, customer1_charges)
print(customer2_tenure, customer2_charges)
print(customer3_tenure, customer3_charges)
````

## Task 2 — High Value, Long-Term Flag

For **each** customer, compute and print whether they are a **"high value, long-term"** customer, defined as:

```
tenure_months > 12   AND   monthly_charges > 70
```

Print a clear, labeled result for each customer (e.g. `"Customer 1 is high value & long-term: True"`).


````python
# Customer 1 check
# High value, long-term means: tenure > 12 AND monthly charges > 70
customer1_flag = customer1_tenure > 12 and customer1_charges > 70
print("Customer 1 is high value & long-term:", customer1_flag)

# Customer 2 check
customer2_flag = customer2_tenure > 12 and customer2_charges > 70
print("Customer 2 is high value & long-term:", customer2_flag)

# Customer 3 check
customer3_flag = customer3_tenure > 12 and customer3_charges > 70
print("Customer 3 is high value & long-term:", customer3_flag)
````

## Task 3 — Bonus: Type Casting

Pick **one** customer's `monthly_charges` and cast it to `int`.

1. Print the original float value.
2. Print the casted int value.
3. In a comment, explain what value was lost and why (hint: `int()` truncates, it does not round).


````python
# Pick Customer 1's monthly charges and cast it to int
print("Original float value:", customer1_charges)

charges_as_int = int(customer1_charges)
print("Casted int value:", charges_as_int)

# int() truncates, it does not round: the .50 after the decimal point
# was simply cut off (lost), so 85.50 became 85, not 86.
````
