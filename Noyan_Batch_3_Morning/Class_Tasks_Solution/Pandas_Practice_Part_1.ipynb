# Setup

Run this cell first — it creates the sample dataset used throughout this notebook's exercises.

````python
import pandas as pd
import numpy as np
import os

os.makedirs("sample_data", exist_ok=True)

sample_csv_text = """name,age,city,department,salary
Ali,28,Lahore,Engineering,85000
Sara,32,Karachi,Marketing,72000
Ahmed,45,Lahore,Engineering,120000
Ayesha,24,Islamabad,Sales,55000
Bilal,38,Karachi,Engineering,98000
Zainab,29,Islamabad,Marketing,68000
Usman,52,Lahore,Sales,60000
Hina,31,Karachi,Engineering,91000
"""

with open("sample_data/employees.csv", "w") as file:
    file.write(sample_csv_text)

print("employees.csv is ready.")
print(pd.__version__)
````

# Level 1 — Basic

### Task 1: Load and Inspect
Read `sample_data/employees.csv` into a DataFrame called `df`. Print the first 3 rows, the shape, and the column data types.

### Task 2: Create a Series
Create a Pandas Series called `department_series` directly from `df["department"]`. Print its `.unique()` values and how many unique values there are.

### Task 3: Select Columns
From `df`, select just the `name` and `salary` columns as a new DataFrame. Print the result.

### Task 4: Row Selection with `.iloc[]`
Print the 4th row of `df` (by position, not label) using `.iloc[]`. Then print just the `city` value of that same row.

### Task 5: Basic Filtering
Filter `df` to show only employees with a `salary` greater than `80000`. Print the result.

````python
# Task 1: Load and Inspect
df = pd.read_csv("sample_data/employees.csv")
print(df.head(3))          # first 3 rows
print("Shape:", df.shape)  # (rows, columns)
print(df.dtypes)           # data type of each column

# Task 2: Create a Series
department_series = df["department"]
print("Unique departments:", department_series.unique())
print("Number of unique departments:", department_series.nunique())

# Task 3: Select Columns
name_salary = df[["name", "salary"]]
print(name_salary)

# Task 4: Row Selection with `.iloc[]`
print("4th row (position 3):")
print(df.iloc[3])
print("City of the 4th row:", df.iloc[3]["city"])

# Task 5: Basic Filtering
high_earners = df[df["salary"] > 80000]
print("Employees with salary > 80000:")
print(high_earners)
````

# Level 2 — Applied

### Task 6: Combined Filtering
Filter `df` to show only employees in the `"Engineering"` department **and** with `age` less than `40`. Print the result.

### Task 7: Add a Computed Column
Add a new column `annual_bonus` to `df` equal to 10% of each employee's `salary`. Print the updated DataFrame.

### Task 8: `.loc[]` for Row + Column Selection
Using `.loc[]`, print only the `name` and `department` columns for employees whose `city` is `"Karachi"`.

### Task 9: Handling Missing Data
Create a small copy of `df` where you manually set the `salary` of two rows to `np.nan` (use `.loc[]` to do this). Then: (a) print how many missing values exist per column, (b) fill the missing `salary` values with the column's average, and print the result.

### Task 10: Sorting
Sort `df` by `department` (ascending) and then by `salary` (descending) within each department. Print the result.

````python
# Task 6: Combined Filtering
# & means AND: both conditions must be True (each condition in parentheses)
engineering_young = df[(df["department"] == "Engineering") & (df["age"] < 40)]
print(engineering_young)
````

````python
# Task 7: Add a Computed Column
# 10% of salary as an annual bonus
df["annual_bonus"] = df["salary"] * 0.10
print(df)
````

````python
# Task 8: `.loc[]` for Row + Column Selection
# .loc[rows, columns]: rows where city is Karachi, only name + department columns
print(df.loc[df["city"] == "Karachi", ["name", "department"]])
````

````python
# Task 9: Handling Missing Data
# Work on a COPY so the original df stays complete for later tasks
df_missing = df.copy()

# Manually set the salary of two rows to NaN, using .loc[]
df_missing.loc[0, "salary"] = np.nan
df_missing.loc[3, "salary"] = np.nan

# (a) How many missing values per column?
print("Missing values per column:")
print(df_missing.isna().sum())

# (b) Fill missing salaries with the column's average (mean ignores NaN)
average_salary = df_missing["salary"].mean()
df_missing["salary"] = df_missing["salary"].fillna(average_salary)
print("After filling missing salaries:")
print(df_missing)
````

````python
# Task 10: Sorting
# Sort by department A-Z, then by salary high-to-low inside each department
sorted_df = df.sort_values(by=["department", "salary"], ascending=[True, False])
print(sorted_df)
````

# Level 3 — ML-oriented Challenge

### Task 11: Feature Engineering
Using `df`, create a new column `senior_employee` that is `True` if `age >= 40`, otherwise `False`. Then use boolean indexing to print only the senior employees, showing just their `name`, `age`, and `department`.

### Task 12: Cleaning a Messier Dataset
Create a new DataFrame from this dictionary:
```python
messy = {
    "name": ["Ali", "Sara", None, "Ayesha", "Bilal"],
    "age": [28, np.nan, 45, 24, 38],
    "salary": [85000, 72000, 120000, np.nan, 98000],
}
```
1. Print how many missing values are in each column.
2. Drop any row where `name` is missing (hint: `df.dropna(subset=["name"])`).
3. Fill any remaining missing `age` or `salary` values with that column's mean.
4. Print the final, cleaned DataFrame.

### Task 13: A Small Exploration Report
Using the original `df` (employees dataset), write code that prints, in a clean format:
- The total number of employees.
- The average salary, rounded to 2 decimal places.
- The department with the highest average salary (hint: you'll need to compute an average salary per department — a preview of `groupby`, which you'll learn properly in Part 2. For now, try filtering + `.mean()` for each department one at a time, or explore if `df.groupby("department")["salary"].mean()` works for you).
- The name of the single highest-paid employee.

````python
# Task 11: Feature Engineering
# True if age >= 40, otherwise False
df["senior_employee"] = df["age"] >= 40

# Show only senior employees, with just name, age, department
print(df.loc[df["senior_employee"], ["name", "age", "department"]])
````

````python
# Task 12: Cleaning a Messier Dataset
messy = {
    "name": ["Ali", "Sara", None, "Ayesha", "Bilal"],
    "age": [28, np.nan, 45, 24, 38],
    "salary": [85000, 72000, 120000, np.nan, 98000],
}
messy_df = pd.DataFrame(messy)

# 1. Missing values per column
print("Missing values per column:")
print(messy_df.isna().sum())

# 2. Drop rows where name is missing
messy_df = messy_df.dropna(subset=["name"])

# 3. Fill remaining missing age/salary with that column's mean
messy_df["age"] = messy_df["age"].fillna(messy_df["age"].mean())
messy_df["salary"] = messy_df["salary"].fillna(messy_df["salary"].mean())

# 4. Final cleaned DataFrame
print("Cleaned DataFrame:")
print(messy_df)
````

````python
# Task 13: A Small Exploration Report
print("Total number of employees:", len(df))
print("Average salary:", round(df["salary"].mean(), 2))

# Department with the highest average salary (groupby preview)
avg_salary_by_dept = df.groupby("department")["salary"].mean()
print("Average salary per department:")
print(avg_salary_by_dept)
print("Department with highest average salary:", avg_salary_by_dept.idxmax())

# Name of the single highest-paid employee
top_row = df.loc[df["salary"].idxmax()]
print("Highest-paid employee:", top_row["name"])
````
